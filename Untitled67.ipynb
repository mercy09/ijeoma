```python
#Load the Dataset
import pandas as pd

df = pd.read_csv("employee_performance_workload_attrition.csv")

df.head()
df.info()
df.describe()
```

```python
df.head(20)
```

```python
#Check for Missing values and Duplicate rows
df.isnull().sum()
df.duplicated().sum()
```

```python
df['attrition'] = df['attrition'].str.strip().str.lower()
df['Attrition'] = df['attrition'].map({'yes': 1, 'no': 0})
```

```python
df.head(20)
```

```python
#Create At Least 8 Visualizations
total_emp = df.shape[0]
attrition_yes = df[df['Attrition']==1].shape[0]
attrition_rate = (attrition_yes / total_emp) * 100
avg_performance = df['performance_rating'].mean()
avg_workload = df['avg_weekly_hours'].mean()
```

```python
summary = {
    "Total Employees": total_emp,
    "Total Attrition (Yes)": attrition_yes,
    "Attrition Rate (%)": round(attrition_rate, 2),
    "Average Performance Score": round(avg_performance, 2),
    "Average Workload Hours": round(avg_workload, 2)
}

summary_df = pd.DataFrame(summary, index=[0])
print(summary_df)
```

```python
#VISUAL 2: Attrition Distribution (Bar Chart)
import seaborn as sns
import matplotlib.pyplot as plt

sns.countplot(x='Attrition', data=df)
plt.title("Attrition Distribution")
plt.show()
```

```python
#VISUAL 3: Workload vs Attrition (Box Plot)
sns.boxplot(x='Attrition', y='avg_weekly_hours', data=df)
plt.title('Workload vs Attrition')
plt.show()
```

```python
#VISUAL 4: Performance Score vs Attrition
sns.boxplot(x='Attrition', y='performance_rating', data=df)
plt.title('Performance Score vs Attrition')
plt.show()
```

```python
#VISUAL 5: Correlation Heatmap
corr = df.corr(numeric_only=True)

sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Matrix")
plt.show()
```

```python
#VISUAL 6: Overtime/Workload Category vs Attrition
df['Workload_Category'] = pd.cut(df['avg_weekly_hours'],
                                  bins=[0,40,60,100],
                                  labels=['Low','Medium','High'])
sns.countplot(x='Workload_Category', hue='Attrition', data=df)
plt.show()
```

```python
#VISUAL 7: Department vs Attrition (if department column exists)
pd.crosstab(df['department'], df['Attrition']).plot(kind='bar')
plt.show()
```

```python
#VISUAL 8: Age vs Attrition (Histogram or Boxplot)
sns.boxplot(x='Attrition', y='avg_weekly_hours', data=df)
plt.show()
```

```python
#Job Role vs Monthly Income (Boxplot)
plt.figure(figsize=(10,5))
sns.boxplot(x='role_level', y='monthly_salary', data=df)
plt.xticks(rotation=45)
plt.title('Job Role vs Monthly Income')
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.scatterplot(x='projects_handled', y='job_satisfaction', hue='Attrition', data=df)
plt.title('Experience vs Work Life Balance')
plt.show()
```

```python
# Average income by department
avg_income_dept = df.groupby('department')['monthly_salary'].mean()
print("Average Income by Department:\n", avg_income_dept)
```

```python
# Employees with low WorkLifeBalance
low_wlb = df[df['job_satisfaction'] <= 2]['employee_id'].count()
print(f"Employees with low Work Life Balance: {low_wlb}")
```

```python
# Attrition rate
attrition_rate = df['Attrition'].value_counts(normalize=True) * 100
print("Attrition Rate:\n", attrition_rate)
```
